#### Importación de librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('dataset_PetFinder.csv')
```

#### Filtrado de registros, Type = 2 (gatos)

```python
df_gatos = df[df["Type"] == 2].copy()
```

```python
df_gatos.shape
```

```python
df_gatos["Type"].value_counts()
```

```python
df_gatos.drop(columns="Type", inplace=True)
```

#### Renombramiento de columnas y exportación de csv

```python
df_gatos = df_gatos.rename(columns={
    "Name": "Nombre",
    "Age": "Edad_Meses",
    "Breed1": "Raza_Principal",
    "Breed2": "Raza_Secundaria",
    "Gender": "Genero",
    "Color1": "Color_Principal",
    "Color2": "Color_Secundario",
    "Color3": "Color_Terciario",
    "MaturitySize": "Tamano_Madurez",
    "FurLength": "Longitud_Pelo",
    "Vaccinated": "Vacunado",
    "Dewormed": "Desparasitado",
    "Sterilized": "Esterilizado",
    "Health": "Salud",
    "Quantity": "Cantidad",
    "Fee": "Cuota_Adopcion",
    "State": "Estado",
    "RescuerID": "ID_Rescatista",
    "VideoAmt": "Cantidad_Videos",
    "Description": "Descripcion",
    "PetID": "ID_Mascota",
    "PhotoAmt": "Cantidad_Fotos",
    "AdoptionSpeed": "Velocidad_Adopcion"
})
```

```python
df_gatos.to_csv("dataset_gatos.csv",index=False,encoding="utf-8-sig")
```

#### Casting semántico

```python
# Casting de variables de texto
columnas_texto = ["Descripcion","ID_Rescatista","ID_Mascota"]
df_gatos[columnas_texto] = df_gatos[columnas_texto].astype("string")

# Casting de variables numéricas
df_gatos["Edad_Meses"] = df_gatos["Edad_Meses"].astype("int16")
df_gatos["Cantidad"] = df_gatos["Cantidad"].astype("int8")
df_gatos["Cuota_Adopcion"] = df_gatos["Cuota_Adopcion"].astype("int16")
df_gatos["Cantidad_Videos"] = df_gatos["Cantidad_Videos"].astype("int8")

# Cantidad_Fotos puede contener NA, por eso usamos Int8 (nullable)
df_gatos["Cantidad_Fotos"] = df_gatos["Cantidad_Fotos"].astype("Int8")

# Casting de variables categóricas
columnas_categoria = [
    "Genero",
    "Tamano_Madurez",
    "Longitud_Pelo",
    "Vacunado",
    "Desparasitado",
    "Esterilizado",
    "Salud"
]

for col in columnas_categoria:
    df_gatos[col] = df_gatos[col].astype("category")

print("Casting realizado correctamente.")
df_gatos.info()
```

#### Etiquetar categorías

```python
df_gatos["Genero"] = df_gatos["Genero"].cat.rename_categories({
    1: "Macho",
    2: "Hembra",
    3: "Mixto"
})
```

```python
df_gatos["Tamano_Madurez"] = df_gatos["Tamano_Madurez"].cat.rename_categories({
    0: "No especificado",
    1: "Pequeño",
    2: "Mediano",
    3: "Grande",
    4: "Extra grande"
})
```

```python
df_gatos["Longitud_Pelo"] = df_gatos["Longitud_Pelo"].cat.rename_categories({
    0: "No especificado",
    1: "Corto",
    2: "Medio",
    3: "Largo"
})
```

```python
dic_si_no = {
    1: "Sí",
    2: "No",
    3: "No seguro"
}

df_gatos["Vacunado"] = df_gatos["Vacunado"].cat.rename_categories(dic_si_no)
df_gatos["Desparasitado"] = df_gatos["Desparasitado"].cat.rename_categories(dic_si_no)
df_gatos["Esterilizado"] = df_gatos["Esterilizado"].cat.rename_categories(dic_si_no)
```

```python
df_gatos["Salud"] = df_gatos["Salud"].cat.rename_categories({
    0: "No especificado",
    1: "Sano",
    2: "Lesión leve",
    3: "Lesión grave"
})
```

```python
# Etiquetar velocidad de adopción
mapa_velocidad = {
    0: "Mismo día",
    1: "1–7 días",
    2: "8–30 días",
    3: "31–90 días",
    4: "No adoptado (+100 días)"
}

df_gatos["Velocidad_Adopcion_Nombre"] = (df_gatos["Velocidad_Adopcion"].map(mapa_velocidad))
```

#### Validación después del casting

```python
# Resumen de tipos de datos
resumen_tipos = pd.DataFrame({
    "Tipo_Dato": df_gatos.dtypes.astype(str),
    "Valores_Unicos": df_gatos.nunique()
})

display(resumen_tipos)
```

#### Revisión de datos nulos y duplicados

```python
df_gatos.isnull().sum()
```

```python
duplicados = df_gatos.duplicated().sum()

print(f"Registros duplicados: {duplicados}")
```

#### Revisión de categorías

En esta sección se revisan las variables categóricas del conjunto de datos para conocer los valores presentes, su frecuencia y detectar posibles categorías "No especificado" antes del proceso de limpieza.

```python
# Verificar resultado
df_gatos[
    ["Velocidad_Adopcion", "Velocidad_Adopcion_Nombre"]
].drop_duplicates().sort_values("Velocidad_Adopcion")
```

```python
# Lista de variables categóricas codificadas
columnas_categoricas = [
    "Genero",
    "Tamano_Madurez",
    "Longitud_Pelo",
    "Vacunado",
    "Desparasitado",
    "Esterilizado",
    "Salud",
    "Velocidad_Adopcion_Nombre"
]

print("="*60)
print("REVISIÓN DE CATEGORÍAS DEL DATASET PETFINDER (GATOS)")
print("="*60)

for columna in columnas_categoricas:
    print(f"\n 🔶 Variable: {columna}")
    print("-"*60)

    categorias = (df_gatos[columna].value_counts(dropna=False).sort_index().rename("Frecuencia").to_frame())

    categorias["Porcentaje (%)"] = (categorias["Frecuencia"] / len(df_gatos) * 100).round(2)
    display(categorias)
```

### Revisión de rangos

En esta sección se analizan las variables numéricas del conjunto de datos para identificar sus rangos, valores mínimos y máximos, posibles valores atípicos y distribuciones generales.

```python
variables_numericas = [
    "Edad_Meses",
    "Cantidad",
    "Cuota_Adopcion",
    "Cantidad_Fotos",
    "Cantidad_Videos"
]

estadisticas = df_gatos[variables_numericas].describe().T
estadisticas = estadisticas[["count","mean","std","min","25%","50%","75%","max"]].round(2)
display(estadisticas)
```

```python
resumen_rangos = pd.DataFrame({
    "Mínimo": df_gatos[variables_numericas].min(),
    "Máximo": df_gatos[variables_numericas].max(),
    "Valores únicos": df_gatos[variables_numericas].nunique()
})

display(resumen_rangos)
```

### Tratamiento a datos nulos

```python
print(f"Nombres nulos: {df_gatos['Nombre'].isna().sum()}")
```

```python
nombres_nulos = df_gatos[df_gatos["Nombre"].isna()]
print(f"Total de registros sin nombre: {len(nombres_nulos)}")
display(
    nombres_nulos[
        ["ID_Mascota", "Edad_Meses", "Genero", "Raza_Principal",
         "Estado", "Cantidad_Fotos", "Descripcion","Nombre"]
    ].head(10)
)
```

```python
df_gatos["Nombre"] = (df_gatos["Nombre"].fillna("Sin nombre").str.strip())
```

```python
descripcion_nula = df_gatos[df_gatos["Descripcion"].isna()]
print(f"Descripciones nulas: {len(descripcion_nula)}")
display(descripcion_nula[["ID_Mascota", "Nombre", "Edad_Meses","Genero", "Raza_Principal", "Cantidad_Fotos"]])
```

```python
df_gatos["Descripcion"] = (df_gatos["Descripcion"].fillna("Sin descripción").str.strip())
```
